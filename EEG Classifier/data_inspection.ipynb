```python
from pathlib import Path
import re
import pandas as pd

DATA_DIR = Path(r"C:\College Work\Project-1\data\raw")

files = list(DATA_DIR.glob("*.edf"))

print("Total EDF files:", len(files))

for file in files[:20]:
    print(file.name)
```

```python
records = []

for file in files:
    name = file.stem

    if name.startswith("H "):
        label = 0
        group = "Healthy"
    elif name.startswith("MDD "):
        label = 1
        group = "MDD"
    else:
        print("Unknown filename:", name)
        continue

    match = re.search(r"S(\d+)", name)

    if match:
        subject_id = int(match.group(1))
    else:
        print("Could not find subject:", name)
        continue

    if "EC" in name:
        condition = "EC"
    elif "EO" in name:
        condition = "EO"
    elif "TASK" in name:
        condition = "TASK"
    else:
        condition = "UNKNOWN"

    records.append({
        "file": file.name,
        "subject_id": subject_id,
        "group": group,
        "label": label,
        "condition": condition
    })

metadata = pd.DataFrame(records)

metadata.head()
```

```python
print("\nGroup counts:")
print(metadata["group"].value_counts())

print("\nCondition counts:")
print(metadata["condition"].value_counts())

print("\nGroup × Condition:")
print(
    pd.crosstab(
        metadata["group"],
        metadata["condition"]
    )
)

print("\nUnique subjects:")
print(
    metadata.groupby("group")["subject_id"]
    .nunique()
)
```

```python
metadata.to_csv(
    r"C:\College Work\Project-1\data\metadata.csv",
    index=False
)

print("Saved metadata.csv")
```
